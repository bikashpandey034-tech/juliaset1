```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-2, 2, 1000)
y = np.linspace(-2, 2, 1000)

X, Y = np.meshgrid(x, y)

Z = X + 1j * Y

c = 0.7 + 0.23525413954j

iterations = np.full(Z.shape, -1)

for i in range(100):

    Z = Z**2 + c

    escaped = (np.abs(Z) > 2) & (iterations == -1)

    iterations[escaped] = i

plt.imshow(
    iterations,
    extent=[-2, 2, -2, 2]
)

plt.xlabel("Real")
plt.ylabel("Imaginary")
plt.title("Julia Set")

plt.show()
```
